In this lab you are going to explore and create a program that defines a class call bottle

```python
class bottle:
  def __init__(self,brand,material,size):
    self.brand=brand
    self.material=material
    self.size=size

  def close(self):
    print("The bottle is closed")
  def open(self):
    print("The bottle is opened")
  def fill(self):
    print("The bottle is filled")
  def describe(self):
    print(f"This bottle was made by {self.brand}, the material is {self.material} & size: {self.size}")


#self se utiliza para decirle a python a que objeto pertenecen los atributos
#Create an instance using class bottle
bottle1 = bottle("Epura","Plastic","20cm3")
bottle2 = bottle("Ciel","Aluminum","25x10cm")

#We access to the object(Instance) "bottle 1" to call its data
print(bottle1.brand)
print(bottle1.material)
print(bottle1.size)
bottle1.fill()
bottle1.describe()

print("-"*70)

#Instance "Bottle 2":
print(bottle2.brand)
print(bottle2.material)
print(bottle2.size)
bottle2.fill()
bottle2.describe()
```

#Lab 1: Bank Account Class
Create a Python program that models a bank account using Object-Oriented Programming

```python
class BankAccount:
  def __init__(self,holdersname,__balance):
    self.holdersname = holdersname
    self.__balance = __balance

  def deposit(self,amount):
    if amount<=0:
      print("Does not allow negative or zero amounts")
    else:
      self.__balance+=amount

  def withdraw(self,money):
    if money>self.__balance:
      print("Does not allow withdrawals greater than available balance")
    else:
      self.__balance-=money

  def check_balance(self):
    print(self.__balance)


count1 = BankAccount("Raúl Pérez",5000)
count2 = BankAccount("Joel López",3000)

print(count1.holdersname)
count1.check_balance()
amount=(float)(input("Ingresa el  monto a ingresar a la cuenta: ").strip())
count1.deposit(amount)
money=(float)(input("Ingresa el  monto a retirar de la cuenta: ").strip())
count1.withdraw(money)
count1.check_balance()
print("-"*75)
print(count2.holdersname)
count2.check_balance()
amount=(float)(input("Ingresa el  monto a ingresar a la cuenta: ").strip())
count2.deposit(amount)
money=(float)(input("Ingresa el  monto a retirar de la cuenta: ").strip())
count2.withdraw(money)
count2.check_balance()
```
